```python
import pandas as pd

prices = pd.read_csv("route_pricing.csv")
det = pd.read_csv("detention_by_route.csv")
print(det.head())    
print(det.shape)     

df = prices.merge(det[["route_id", "avg_detention_min_per_load"]], on="route_id")
print(df.shape)      

fair_price = df["billed_per_mile"].quantile(0.25)
df["flagged"] = df["billed_per_mile"] < fair_price

df.groupby("flagged")["avg_detention_min_per_load"].agg(["count", "mean", "min", "max"])
```
